
**Question 1: Extracting Elements with Conditions (Boolean Masking)   Create a 4X4 matrix containing numbers from 1 to 16. Extract all elements that are greater than 10 and divisible by 2.**


```python
import numpy as np

# Create a 4x4 matrix with values from 1 to 16
matrix = np.arange(1, 17).reshape(4, 4)

print("Matrix:")
print(matrix)


result = matrix[(matrix > 10) & (matrix % 2 == 0)]

print("Elements greater than 10 and divisible by 2:")
print(result)
```


**QUESTION 2 : SIDE-BY-SIDE SUBPLOTS 	CREATE A FIGURE THAT CONTAINS TWO PLOTS SIDE BY SIDE(1 ROW ,2 COLUMN)		LEFT PLOT: A LINE POLOT SHOWING HOW TRAINIG ACCURACY INCREASE OVER 5 EPOCHS		RIGHT PLOT : A BAR CHART DISPLAYING THE INFERENCE SPEED (FRAMES PER SECOND ) OF 3 DIFFERENT HARDWARE SETUPS**


```python
import matplotlib.pyplot as plt

# Data
epochs = [1, 2, 3, 4, 5]
accuracy = [60, 68, 75, 82, 90]

hardware = ['CPU', 'GPU', 'TPU']
fps = [30, 120, 250]

# Create a figure with 1 row and 2 columns
plt.figure(figsize=(10, 4))

# Left subplot: Line plot
plt.subplot(1, 2, 1)
plt.plot(epochs, accuracy, marker='o', color='blue')
plt.title("Training Accuracy Over 5 Epochs")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.grid(True)

# Right subplot: Bar chart
plt.subplot(1, 2, 2)
plt.bar(hardware, fps, color=['red', 'green', 'orange'])
plt.title("Inference Speed on Different Hardware")
plt.xlabel("Hardware")
plt.ylabel("Frames Per Second (FPS)")

# Adjust layout and display
plt.tight_layout()
plt.show()
```
